```python
import seaborn as sns
import matplotlib.pyplot as plt

# Boxplot to detect outliers
sns.boxplot(x=df['Cart_Value'])
plt.show()
```

```python
list1=['Browsing_Time_Minutes','Items_Viewed','Cart_Value']
for col in list1:
    sns.boxplot(x= df[col])
    plt.show()
```

```python
Q1 = df['Cart_Value'].quantile(0.25)
Q3 = df['Cart_Value'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Filter out outliers
df = df[(df['Cart_Value'] >= lower_bound) &
         (df['Cart_Value'] <= upper_bound)]
```

```python
sns.boxplot(x=df['Cart_Value'])
plt.show()
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
df[['Age_scaled', 'Browsing_Time_scaled']] = scaler.fit_transform(df[['Age', 'Browsing_Time_Minutes']])
df[['Age_scaled', 'Browsing_Time_scaled']].head()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['Cart_Value'],bins=30)
plt.title("Distribution of Cart Value")
plt.xlabel("Cart Value")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x='Age', y='Cart_Value', hue='Gender', data=df)
plt.title("Age vs Cart Value")
plt.show()
```

```python
from scipy.stats import mannwhitneyu
u_stat, p_val_u = mannwhitneyu(df['Age'], df['Cart_Value'])
print(f"Mann-Whitney U Test p-value: {p_val_u}")
```

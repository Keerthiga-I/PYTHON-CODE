```python
import random
import numpy as np
import pandas as pd
```

```python
np.random.seed(60)
```

```python
n=1000
```

```python
milk= np.random.choice([1,0],size=n,p=[0.6,0.4])
bread = np.random.choice([1,0],size=n,p=[0.5,0.5])
```

```python
print(milk)
```

```python
df=pd.DataFrame({'Milk':milk,'Bread':bread})
print(df)
```

```python
#Marginal Prob
#Milk
p_milk = df['Milk'].mean()
print(p_milk)
```

```python
p_bread = df['Bread'].mean()
print(p_bread)
```

```python
#Joint Prob
m_and_b = ((df['Milk']==1) &(df['Bread']==1)).mean()
print(m_and_b)
```

```python
#Union Prob
m_or_b = ((p_milk + p_bread-m_and_b))
print(m_or_b)
```

```python
#Conditional Prob
p_b_given_m = m_and_b/p_milk
print(f'{p_b_given_m:.2f}')
```

```python
p_br_given_mi = df[df['Milk']==1]['Bread'].mean()
print(p_br_given_mi)
```

```python
p_mi_given_br = df[df['Bread']==1]['Milk'].mean()
print(p_mi_given_br)
```

```python
p_m_given_b = m_and_b/p_bread
print(f'{p_m_given_b:.2f}')
```

```python
#Visualisation
import matplotlib.pyplot as plt

df.groupby(['Milk','Bread']).size().unstack().plot(kind='bar',stacked=True)
plt.title('Purchase Pattern of Milk and Bread')
plt.xlabel('Milk Purchase(0=No,1=Yes)')
plt.ylabel('Number Of Customer')
plt.legend(['Bread = No','Bread = Yes'])
plt.show()
```

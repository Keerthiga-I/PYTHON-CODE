```python
lst = [x for x in range(6)]
lst
# start end and step
lst[1:3:1]
```

```python

# default step size is 1
lst[1:5]
```

```python
# sushi operator
lst[::2]
```

```python

# reversing the list order
lst[::-1]
```

```python
# deleting the contents of a list but not the object
lst = [1,2,3,4,5]
del lst[:]
lst
```

```python
# equivalent
lst = [1,2,3,4,5]
lst.clear()
lst
```

```python
# replace all elements without creating a new list object
original_lst = lst
lst[:] = [7, 8, 9]
print(lst)
print(original_lst)
original_lst.append(11)
print(lst)
print(original_lst)
```

```python
#### Shallow copies
copied_lst = lst[:]
copied_lst is lst
copied_lst.append(10)
print(copied_lst)
print(lst)
```

```python
### list comprehensions vs generator expressions
listcomp = ['Hello' for i in range(3)]
genexpr = ('Hello' for i in range(3))
print(listcomp)
```

```python
print(genexpr)
```

```python
list(genexpr)
```

```python
even_squares = [x * x for x in range(10) if  x % 2==0]
for x in even_squares:
    print(x)
```

```python
### In-line Generator Expressions
##You can create an iterator and consume it straight away

##You can create an iterator and consume it straight away
for num in (x*x for x in range(10) if x % 2 == 0):
    print(num)
```

```python
for x in ('Bom dia' for i in range(3)):
  print(x)
```

```python
sum((x*2 for x in range(10)))
```
